```python
import numpy as np
import matplotlib.pyplot as plt


#设定中文
plt.rcParams['font.sans-serif'] = ['SimHei']  # 用来正常显示中文标签
plt.rcParams['axes.unicode_minus'] = False  # 用来正常显示负号

# 1. 设定参数
FA0 = 10.0
CA0 = 1.0
k = 0.5
X_target = 0.8

# 2. 计算 PFR 达到 80% 转化率的体积
# 提示：V = (FA0 / (k * CA0)) * np.log(1 / (1 - X_target))
V_pfr_target = (FA0 / (k * CA0)) * np.log(1 / (1 - X_target))
print(f"达到 80% 转化率所需的 PFR 体积为: {V_pfr_target:.2f} L")

# 3. 生成转化率数组和体积数组进行对比
X_array = np.linspace(0.01, 0.99, 100)

# CSTR 体积数组 (复习一下，分母是 1 - X_array)
V_cstr_array = (FA0 * X_array) / (k * CA0 * (1 - X_array))

# PFR 体积数组
# 提示：V_pfr_array = (FA0 / (k * CA0)) * np.log(1 / (1 - X_array))
V_pfr_array = (FA0 / (k * CA0)) * np.log(1 / (1 - X_array))

# 4. 绘图
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(V_cstr_array, X_array, color='tab:red', linewidth=2, label='CSTR (全混流)')
ax.plot(V_pfr_array, X_array, color='tab:blue', linewidth=2, label='PFR (平推流)')

ax.set_xlim(0, 800) # 限制 X 轴在 800L 以内，方便看清 80% 转化率的区域
ax.set_ylim(0, 1.0)
ax.set_xlabel('反应器体积 V (L)', fontsize=12)
ax.set_ylabel('转化率 X', fontsize=12)
ax.set_title('CSTR 与 PFR 体积对比', fontsize=14)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### 热失控（热点）警告
> - 虽然PFR体积小，但是如果这是强放热反应，PFR入口处由于浓度极高，反应剧烈，会瞬间产生巨大热量，导致“飞温”
> - 而CSTR因为有搅拌，温度分布均匀，反而更安全
> * 所以必须加入温度检测，不要因为体积小就选择PFR
