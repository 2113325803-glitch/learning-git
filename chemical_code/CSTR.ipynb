```python
import numpy as np
import matplotlib.pyplot as plt

#设定中文
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# 1. 设定已知参数
FA0 = 10.0     # 进料流量 mol/s
CA0 = 1.0      # 进料浓度 mol/L
k = 0.1        # 反应速率常数 1/s
X_target = 0.8 # 目标转化率 80%

# 2. 计算达到目标转化率所需的体积
# 提示：V = (FA0 * X_target) / (k * CA0 * (1 - X_target))
V_target = (FA0 * X_target) / (k * CA0 * (1 - X_target))
print(f"达到 80% 转化率所需的 CSTR 体积为: {V_target:.2f} L")

# 3. 进阶：画一条曲线，看看体积和转化率的关系
# 提示：生成 0.01 到 0.99 的转化率数组
X_array = np.linspace(0.01, 0.99, 100)

# 提示：计算每个 X 对应的 V
V_array = (FA0 * X_array) / (k * CA0 * (1 - X_array))

# 4. 用 OO API 画图
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(V_array, X_array, color='tab:blue', linewidth=2)
ax.set_xlabel('反应器体积 V (L)', fontsize=12)
ax.set_ylabel('转化率 X', fontsize=12)
ax.set_title('CSTR 转化率与体积的关系', fontsize=14)
ax.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

#### 量纲统一
> 1. 量纲统一是化学工程中一个重要的概念，它涉及到将不同单位的量进行统一，以便进行比较和计算。
>  - 例如，温度可以使用摄氏度、华氏度或开尔文进行表示，但为了进行比较和计算，通常需要将它们统一为开尔文。
>  - 体积可以使用升、毫升或立方米进行表示，但为了进行比较和计算，通常需要将它们统一为立方米。
>  - 质量可以使用克、千克或吨进行表示，但为了进行比较和计算，通常需要将它们统一为千克。

#### 实际化工厂问题
> 1. 温度波动、停留时间、设备安全余量
> 2. 催化剂，当k=0.1是 V=400L,当k=0.3是 V=80L，所以不惜重金来购买催化剂，可以大大减小体积

#### 收益递减
> 想从 0% 转化到 80%，可能只需要 400 L 的反应器；但想从 80% 提升到 99%，反应器体积可能得扩大到几万升。多出来的体积，在工业上全是白花花的银子和占地成本。所以，你完全懂了为什么化工厂不能盲目追求 100% 转化率。
