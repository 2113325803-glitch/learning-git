```python
import pandas as pd
import numpy as np

# 1. 模拟一个每秒记录一次的 DCS 数据（持续 10 分钟，共 600 行）
# 提示：使用 pd.date_range 生成时间，freq='1s'（每秒一次）
# 模拟温度数据：在 300K 附近随机波动
time_index = pd.date_range('2026-10-16 08:00:00', periods=600, freq='1s')
temp_data = np.random.normal(300, 0.5, 600) # 正态分布随机数
df = pd.DataFrame({'时间': time_index, '温度(K)': temp_data})

# 2. 将“时间”列设置为 DataFrame 的索引
# 提示：使用 df.set_index('时间', inplace=True)
df = df.set_index('时间')
print("原始数据形状（行, 列）:", df.shape) # 应该是 (600, 1)

# 3. 按 5 分钟重采样，求平均值
# 提示：使用 df.resample('5min').mean()
df_5min = df.resample('5min').mean()
print("重采样后数据形状:", df_5min.shape) # 应该是 (3, 1)，因为 10 分钟刚好 2 个 5 分钟... 等等，算一下 600 秒是 10 分钟，所以是 2 个 5 分钟？还是 3 个？
print(df_5min)

```
